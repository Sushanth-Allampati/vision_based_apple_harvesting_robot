```python
# ============================================================
# CELL 1 — IMPORT LIBRARIES
# ============================================================

import os
from PIL import Image
from IPython.display import display

print("Libraries imported successfully.")
```

```python
# ============================================================
# CELL 2 — SET SAVED GRAPHS FOLDER
# ============================================================

graphs_path = r"C:\Users\Lenovo\OneDrive\Documents\Robotics_projectt\Robotics_project\trained_models\apple_ripeness_model"

print("Graphs folder:")
print(graphs_path)

print()
print("Folder exists:", os.path.isdir(graphs_path))
```

```python
# ============================================================
# CELL 3 — LIST SAVED GRAPH FILES
# ============================================================

graph_files = [
    "results.png",
    "confusion_matrix.png",
    "confusion_matrix_normalized.png",
    "BoxP_curve.png",
    "BoxR_curve.png",
    "BoxPR_curve.png",
    "BoxF1_curve.png"
]

print("Checking saved graphs...")
print()

for graph_name in graph_files:

    graph_path = os.path.join(
        graphs_path,
        graph_name
    )

    print(
        f"{graph_name:<35} "
        f"Exists: {os.path.isfile(graph_path)}"
    )
```

```python
# ============================================================
# CELL 4 — DISPLAY TRAINING RESULTS GRAPH
# ============================================================

results_graph_path = os.path.join(
    graphs_path,
    "results.png"
)

if os.path.isfile(results_graph_path):

    results_graph = Image.open(results_graph_path)

    print("Training results graph:")
    print(results_graph_path)
    print()

    display(results_graph)

else:
    print("results.png was not found.")
```

```python
# ============================================================
# CELL 5 — DISPLAY CONFUSION MATRIX
# ============================================================

confusion_matrix_path = os.path.join(
    graphs_path,
    "confusion_matrix.png"
)

if os.path.isfile(confusion_matrix_path):

    confusion_matrix = Image.open(
        confusion_matrix_path
    )

    print("Confusion matrix:")
    print(confusion_matrix_path)
    print()

    display(confusion_matrix)

else:
    print("confusion_matrix.png was not found.")
```
